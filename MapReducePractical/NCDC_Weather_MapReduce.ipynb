## Task 2: Implement the Mapper

The Mapper reads each weather record, extracts the year and temperature, and generates intermediate key-value pairs in the form:

(year, temperature)

```python
def mapper(line):
    parts = line.split()

    if len(parts) == 3:
        station, year, temperature = parts
        return year, int(temperature)

    return None


with open("weather.txt", "r") as file:
    for line in file:
        result = mapper(line.strip())

        if result:
            print(result)
```

## Task 3: Implement the Reducer

The Reducer groups all temperature values belonging to the same year and calculates the maximum temperature recorded for that year.

```python
from collections import defaultdict

year_temperatures = defaultdict(list)

with open("weather.txt", "r") as file:
    for line in file:
        result = mapper(line.strip())

        if result:
            year, temperature = result
            year_temperatures[year].append(temperature)

print("Reducer Output:")

for year in sorted(year_temperatures):
    maximum = max(year_temperatures[year])
    print(year, maximum)
```

```python
with open("max_temperature_output.txt", "w") as file:
    for year in sorted(year_temperatures):
        maximum = max(year_temperatures[year])
        file.write(f"{year}\t{maximum}\n")

print("Output file created successfully.")
```

## Task 5: Analyze Yearly Maximum Temperature

The yearly maximum temperatures generated by the Reducer are analyzed to identify the year with the highest recorded temperature.

```python
year_maximums = {}

with open("max_temperature_output.txt", "r") as file:
    for line in file:
        year, temperature = line.split()
        year_maximums[year] = int(temperature)

hottest_year = max(year_maximums, key=year_maximums.get)
highest_temperature = year_maximums[hottest_year]

print("Yearly Maximum Temperatures:")
for year, temperature in year_maximums.items():
    print(year, temperature)

print("\nHottest Year:", hottest_year)
print("Highest Temperature:", highest_temperature)
```

## Supplementary Problem 1: Minimum Temperature for Each Year

```python
print("Minimum Temperature for Each Year:")

for year in sorted(year_temperatures):
    minimum = min(year_temperatures[year])
    print(year, minimum)
```

## Supplementary Problem 2: Average Yearly Temperature

```python
print("Average Temperature for Each Year:")

for year in sorted(year_temperatures):
    average = sum(year_temperatures[year]) / len(year_temperatures[year])
    print(year, round(average, 2))
```

## Supplementary Problem 3: Highest Temperature Recorded Overall

```python
all_temperatures = []

for temperatures in year_temperatures.values():
    all_temperatures.extend(temperatures)

highest = max(all_temperatures)

print("Highest Temperature Overall:", highest)

for year, temperatures in year_temperatures.items():
    if highest in temperatures:
        print("Recorded in Year:", year)
```
